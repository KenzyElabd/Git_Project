```python
import PyPDF2
```

```python
Pdf=open("CV_Kenzy.pdf","rb")
```

```python
PdfReader=PyPDF2.PdfReader(Pdf)
count=len(PdfReader.pages)
textlist=[]
```

```python
for i in range(count):
  try :
    page=PdfReader.pages[i] # Changed from getPage to pages[i] for PyPDF2 v3.0.0+
    textlist.append(page.extract_text()) # Changed from extractText to extract_text() for PyPDF2 v3.0.0+
  except:
    pass
```

```python
textString=" ".join(textlist)
print(f"Content of textString: {textString[:200]}...") # Print a snippet for verification
```

```python
language="en"
```

```python
textString=" ".join(textlist) # Ensure textString is up-to-date with the extracted text
Audio=gTTS(text=textString,lang=language,slow=False)
```

```python
Audio.save('output.mp3')
print("Audio saved as output.mp3")
```
